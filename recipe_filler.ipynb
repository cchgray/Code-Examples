```python
import pandas as pd
import numpy as np

import itertools
```

```python
#@title Default title text


inFileName = "" #@param {type:"string"}
```

```python
#@title Default title text


outFileName = "" #@param {type:"string"}
```

```python
#inFileName = 'Book2.csv'
colnames=['Identifier', 'Formula	Calc Val',	'Override Val',	'Description',	'UOM', 'Override Lo Lim',	'Override Hi Lim',	'Verify']


inDf = pd.read_csv(inFileName, names=colnames, header=None) #this is for Appendix D Sheet
```

```python
inDf
```

```python
currIndex = 0
ids = []
forms = []
calcs = []
ovrs = []
descs= []
uns = []
lolims = []
hilims = []
vers = []
regs = []

for count, (id, form, calc, ovr, desc, un, lolim, hilim, ver) in enumerate(zip(inDf['Identifier'],
                                                                               inDf['Formula'],inDf['Calc Val'],inDf['Override Val'],inDf['Description'],
                                                                               inDf['UOM'],inDf['Override Lo Lim'],inDf['Override Hi Lim'],inDf['Verify'])):

    #print(currIndex)

    if currIndex == int(str(id)[str(id).find('F_')+2:]):

        ids.append(id)
        forms.append(form)
        calcs.append(calc)
        ovrs.append(ovr)
        if currIndex < 100:
            descs.append(str(desc[:desc.rindex(' ')] + '(' + desc[desc.rindex(' ')+1:] + ')'))
        else:
            descs.append(str(desc))
        uns.append(un)
        lolims.append(lolim)
        hilims.append(hilim)
        vers.append(ver)

        regs.append('F95[' + str(currIndex) + ']')
        currIndex += 1

    else:

        while currIndex <= int(str(id)[str(id).find('F_')+2:]):

            #print(currIndex, str(id)[:str(id).find('F_')+2] + str(currIndex))
            if currIndex == int(str(id)[str(id).find('F_')+2:]):
                ids.append(id)
                forms.append(form)
                calcs.append(calc)
                ovrs.append(ovr)
                if currIndex < 100:
                    descs.append(str(desc[:desc.rindex(' ')] + '(' + desc[desc.rindex(' ')+1:] + ')'))
                else:
                    descs.append(str(desc))
                uns.append(un)
                lolims.append(lolim)
                hilims.append(hilim)
                vers.append(ver)

                regs.append('F95[' + str(currIndex) + ']')
                currIndex += 1

            else:
                ids.append(str(id)[:str(id).find('F_')+2] + str(currIndex))
                forms.append('0')
                calcs.append('0')
                ovrs.append('')
                descs.append(str(id)[:str(id).find('F_')+2] + str(currIndex))
                uns.append('')
                lolims.append('UNRESTRICTED')
                hilims.append('UNRESTRICTED')
                vers.append('ON')

                currIndex += 1


```

```python
for reg in regs:
    print(reg)
```

```python
outDf = pd.DataFrame()

outDf['Identifier '] = ids
outDf['Formula'] = forms
outDf['Calc Val'] = calcs
outDf['Override Val'] = ovrs
outDf['Description'] = descs
outDf['UOM'] = uns
outDf['Override Lo Lim'] = lolims
outDf['Override Hi Lim'] = hilims
outDf['Verify'] = vers

```

```python
outDf.to_csv(outFileName, index = False)
```
